````python
import os
import re
import json
import google.generativeai as genai
from typing import List, Dict, Tuple, Optional
import pandas as pd
from collections import Counter, defaultdict
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import time
import warnings
warnings.filterwarnings('ignore')

class MedicalSLUAssistant:
    """
    Comprehensive Medical Assistant with SLU capabilities
    Extracts symptoms, diagnoses, medications from clinical dialogues
    """

    def __init__(self, gemini_api_key: str):
        self.gemini_api_key = gemini_api_key
        self.initialize_components()
        self.conversation_history = []
        self.patient_records = defaultdict(dict)

    def initialize_components(self):
        """Initialize all components of the medical assistant"""
        # Initialize Gemini
        genai.configure(api_key=self.gemini_api_key)
        self.gemini_model = genai.GenerativeModel('gemini-2.5-flash')

        # Medical knowledge base
        self.medical_ontology = self.load_medical_ontology()

        print("🏥 Medical SLU Assistant Initialized Successfully!")
        print("🔧 Components: Gemini API, Medical Ontology")

    def load_medical_ontology(self) -> Dict:
        """Load medical terminology and relationships"""
        return {
            'symptoms': {
                'cardiovascular': ['chest pain', 'palpitations', 'shortness of breath', 'edema'],
                'respiratory': ['cough', 'wheezing', 'shortness of breath', 'chest congestion'],
                'gastrointestinal': ['nausea', 'vomiting', 'abdominal pain', 'diarrhea', 'constipation'],
                'neurological': ['headache', 'dizziness', 'numbness', 'tingling', 'vision changes'],
                'musculoskeletal': ['joint pain', 'back pain', 'muscle weakness', 'swelling'],
                'general': ['fever', 'fatigue', 'weight loss', 'weight gain', 'loss of appetite']
            },
            'diagnoses': {
                'common': ['hypertension', 'diabetes', 'asthma', 'arthritis', 'anxiety', 'depression'],
                'cardiovascular': ['angina', 'myocardial infarction', 'heart failure', 'arrhythmia'],
                'respiratory': ['pneumonia', 'bronchitis', 'COPD', 'asthma exacerbation'],
                'infectious': ['influenza', 'COVID-19', 'urinary tract infection', 'sinusitis']
            },
            'medications': {
                'analgesics': ['aspirin', 'ibuprofen', 'acetaminophen', 'naproxen'],
                'antibiotics': ['amoxicillin', 'azithromycin', 'doxycycline', 'cephalexin'],
                'cardiovascular': ['lisinopril', 'metoprolol', 'atorvastatin', 'amlodipine'],
                'endocrine': ['metformin', 'insulin', 'levothyroxine', 'glipizide']
            }
        }

    def extract_medical_entities(self, text: str) -> Dict:
        """
        Extract medical entities using Gemini with advanced prompting
        """
        prompt = f"""
        You are a medical SLU specialist. Extract structured medical information from this clinical dialogue.

        TASK: Extract symptoms, diagnoses, medications, and clinical intent with high precision.

        DIALOGUE:
        {text}

        EXTRACTION REQUIREMENTS:

        1. SYMPTOMS: Extract ALL symptoms with attributes:
           - name (standard medical terminology)
           - duration (if mentioned)
           - severity (mild/moderate/severe if mentioned)
           - location (body part if specified)
           - timing (constant/intermittent/acute/chronic)

        2. DIAGNOSES: Extract medical conditions discussed:
           - diagnosis name
           - certainty (confirmed/suspected/ruled_out)
           - status (new/existing/recurrent)

        3. MEDICATIONS: Extract ALL medications:
           - drug name
           - dosage
           - frequency
           - route (oral/topical/injection)
           - purpose
           - status (prescribed/current/stopped)

        4. CLINICAL INTENT: Classify the conversation purpose:
           - new_problem, follow_up, medication_refill, test_results, preventive_care

        5. VITAL SIGNS: Extract numerical values:
           - blood_pressure, heart_rate, temperature, weight, height

        6. RECOMMENDATIONS: Extract doctor's advice and next steps

        Return ONLY valid JSON with this structure:
        {{
            "clinical_intent": "intent_category",
            "symptoms": [
                {{
                    "name": "symptom_name",
                    "duration": "duration",
                    "severity": "severity",
                    "location": "location",
                    "timing": "timing",
                    "context": "sentence_where_mentioned"
                }}
            ],
            "diagnoses": [
                {{
                    "name": "diagnosis_name",
                    "certainty": "certainty",
                    "status": "status",
                    "context": "sentence_where_mentioned"
                }}
            ],
            "medications": [
                {{
                    "name": "medication_name",
                    "dosage": "dosage",
                    "frequency": "frequency",
                    "route": "route",
                    "purpose": "purpose",
                    "status": "status",
                    "context": "sentence_where_mentioned"
                }}
            ],
            "vital_signs": {{
                "blood_pressure": "value",
                "heart_rate": "value",
                "temperature": "value",
                "weight": "value",
                "height": "value"
            }},
            "recommendations": [
                {{
                    "type": "test/treatment/referral/follow_up",
                    "description": "recommendation_details",
                    "timeline": "when_to_do"
                }}
            ],
            "conversation_summary": "brief_clinical_summary"
        }}
        """

        try:
            response = self.gemini_model.generate_content(prompt)
            response_text = response.text.strip()

            # Extract JSON from response
            if '```json' in response_text:
                response_text = response_text.split('```json')[1].split('```')[0]
            elif '```' in response_text:
                response_text = response_text.split('```')[1]

            entities = json.loads(response_text)
            return entities

        except Exception as e:
            print(f"Error in entity extraction: {e}")
            return self.rule_based_fallback(text)

    def rule_based_fallback(self, text: str) -> Dict:
        """Rule-based fallback extraction"""
        entities = {
            "clinical_intent": "unknown",
            "symptoms": [],
            "diagnoses": [],
            "medications": [],
            "vital_signs": {},
            "recommendations": [],
            "conversation_summary": "Fallback extraction used"
        }

        # Simple pattern matching
        text_lower = text.lower()

        # Symptom patterns
        symptom_patterns = {
            'chest pain': r'chest\s+(?:pain|discomfort|pressure)',
            'headache': r'headache|head\s+pain',
            'fever': r'fever|temperature|hot',
            'cough': r'cough|coughing',
            'shortness of breath': r'shortness of breath|difficulty breathing|breathless',
            'nausea': r'nausea|sick to stomach|queasy',
            'fatigue': r'fatigue|tired|exhausted'
        }

        for symptom, pattern in symptom_patterns.items():
            if re.search(pattern, text_lower):
                entities["symptoms"].append({
                    "name": symptom,
                    "duration": "",
                    "severity": "",
                    "location": "",
                    "timing": "",
                    "context": "pattern_match"
                })

        return entities

    def generate_soap_note(self, entities: Dict, patient_context: Dict = None) -> str:
        """Generate comprehensive SOAP note"""
        soap_prompt = f"""
        Generate a professional SOAP note based on the extracted medical information.

        EXTRACTED MEDICAL DATA:
        {json.dumps(entities, indent=2)}

        PATIENT CONTEXT:
        {json.dumps(patient_context or {}, indent=2)}

        Create a detailed SOAP note with these sections:

        SUBJECTIVE:
        - Chief Complaint: Summarize primary concerns
        - History of Present Illness: Detailed symptom timeline
        - Review of Systems: Relevant positive and negative findings
        - Past Medical History: Context from patient history
        - Current Medications: Including new prescriptions
        - Allergies: If mentioned

        OBJECTIVE:
        - Vital Signs: Document all mentioned measurements
        - Physical Exam Findings: Based on conversation
        - Diagnostic Results: Any tests mentioned or ordered

        ASSESSMENT:
        - Problem List: Numbered list of diagnoses and issues
        - Clinical Impressions: Reasoning behind diagnoses
        - Differential Diagnosis: If applicable

        PLAN:
        - Medications: Specific prescriptions with instructions
        - Diagnostics: Tests ordered with rationale
        - Referrals: Specialist referrals if needed
        - Follow-up: Specific follow-up instructions
        - Patient Education: Key points discussed

        Format professionally with clear section headers.
        """

        try:
            response = self.gemini_model.generate_content(soap_prompt)
            return response.text.strip()
        except Exception as e:
            return f"Error generating SOAP note: {e}"

    def analyze_conversation_structure(self, text: str) -> Dict:
        """Analyze dialogue structure and participant roles"""
        lines = [line.strip() for line in text.split('\n') if line.strip()]

        doctor_turns = []
        patient_turns = []

        for line in lines:
            if line.lower().startswith(('doctor:', 'dr:', 'physician:')):
                doctor_turns.append(re.sub(r'^(Doctor|Dr|Physician):\s*', '', line, flags=re.IGNORECASE))
            elif line.lower().startswith(('patient:', 'pt:')):
                patient_turns.append(re.sub(r'^(Patient|Pt):\s*', '', line, flags=re.IGNORECASE))

        # Analyze conversation phases
        conversation_phases = self.identify_conversation_phases(doctor_turns, patient_turns)

        return {
            "total_turns": len(lines),
            "doctor_turns": len(doctor_turns),
            "patient_turns": len(patient_turns),
            "conversation_phases": conversation_phases,
            "doctor_questions": [turn for turn in doctor_turns if '?' in turn],
            "patient_concerns": patient_turns[:3] if patient_turns else []
        }

    def identify_conversation_phases(self, doctor_turns: List[str], patient_turns: List[str]) -> Dict:
        """Identify different phases of clinical conversation"""
        phases = {
            "opening": patient_turns[0] if patient_turns else "",
            "history_taking": ' '.join(patient_turns[:min(3, len(patient_turns))]),
            "examination": ' '.join(doctor_turns[1:min(4, len(doctor_turns))]),
            "diagnosis": ' '.join(doctor_turns[-min(3, len(doctor_turns)):]),
            "treatment_plan": ' '.join(doctor_turns[-min(2, len(doctor_turns)):])
        }
        return phases

    def create_structured_record(self, conversation_id: str, entities: Dict, soap_note: str) -> Dict:
        """Create structured patient record"""
        record = {
            "conversation_id": conversation_id,
            "timestamp": datetime.now().isoformat(),
            "clinical_intent": entities.get("clinical_intent", "unknown"),
            "problems": {
                "symptoms": entities.get("symptoms", []),
                "diagnoses": entities.get("diagnoses", [])
            },
            "treatments": {
                "medications": entities.get("medications", []),
                "recommendations": entities.get("recommendations", [])
            },
            "measurements": entities.get("vital_signs", {}),
            "soap_note": soap_note,
            "summary": entities.get("conversation_summary", "")
        }

        return record

    def generate_clinical_insights(self, patient_records: Dict) -> Dict:
        """Generate clinical insights from multiple conversations"""
        all_symptoms = []
        all_diagnoses = []
        all_medications = []

        for record in patient_records.values():
            all_symptoms.extend([s['name'] for s in record['problems']['symptoms']])
            all_diagnoses.extend([d['name'] for d in record['problems']['diagnoses']])
            all_medications.extend([m['name'] for m in record['treatments']['medications']])

        insights = {
            "symptom_patterns": dict(Counter(all_symptoms).most_common(10)),
            "common_diagnoses": dict(Counter(all_diagnoses).most_common(8)),
            "frequent_medications": dict(Counter(all_medications).most_common(8)),
            "visit_statistics": {
                "total_visits": len(patient_records),
                "common_intents": Counter([r['clinical_intent'] for r in patient_records.values()]),
                "average_symptoms_per_visit": len(all_symptoms) / len(patient_records) if patient_records else 0
            }
        }

        return insights

class MedicalAssistantInterface:
    """User interface for the medical assistant"""

    def __init__(self, slu_assistant: MedicalSLUAssistant):
        self.assistant = slu_assistant
        self.current_patient = None

    def start_session(self):
        """Start a new medical assistant session"""
        print("\n" + "="*60)
        print("🏥 MEDICAL SLU ASSISTANT - CLINICAL DIALOGUE ANALYSIS")
        print("="*60)

        self.current_patient = input("Enter Patient ID or Name: ").strip()
        if not self.current_patient:
            self.current_patient = f"patient_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

        print(f"\n👤 Session started for: {self.current_patient}")
        self.main_menu()

    def main_menu(self):
        """Main interaction menu"""
        while True:
            print("\n📋 MAIN MENU:")
            print("1. 📝 Process Single Text File")
            print("2. 📁 Process Multiple Text Files (Batch)")
            print("3. 📊 View Patient Records")
            print("4. 📈 Clinical Analytics")
            print("5. 🚪 Exit")

            choice = input("\nSelect option (1-5): ").strip()

            if choice == '1':
                self.process_single_file()
            elif choice == '2':
                self.process_multiple_files()
            elif choice == '3':
                self.view_patient_records()
            elif choice == '4':
                self.show_clinical_analytics()
            elif choice == '5':
                print("Thank you for using Medical SLU Assistant!")
                break
            else:
                print("Invalid option. Please try again.")

    def process_single_file(self):
        """Process a single conversation file"""
        print("\n📝 PROCESS SINGLE TEXT FILE")
        file_path = input("Enter the path to the conversation text file: ").strip()

        if not os.path.exists(file_path):
            print("❌ File not found. Please check the path.")
            return

        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                conversation = f.read().strip()

            if not conversation:
                print("❌ File is empty.")
                return

            print(f"✅ Loaded conversation from: {os.path.basename(file_path)}")
            print(f"📏 Conversation length: {len(conversation)} characters")

            self.process_conversation(conversation, os.path.basename(file_path))

        except Exception as e:
            print(f"❌ Error reading file: {e}")

    def process_multiple_files(self):
        """Process multiple conversation files in batch"""
        print("\n📁 PROCESS MULTIPLE TEXT FILES")
        folder_path = input("Enter the path to folder containing text files: ").strip()

        if not os.path.exists(folder_path):
            print("❌ Folder not found. Please check the path.")
            return

        # Get all text files in the folder
        text_files = [f for f in os.listdir(folder_path) if f.endswith('.txt')]

        if not text_files:
            print("❌ No text files found in the folder.")
            return

        print(f"📂 Found {len(text_files)} text files:")
        for i, file in enumerate(text_files, 1):
            print(f"   {i}. {file}")

        confirm = input("\nProcess all files? (y/n): ").strip().lower()
        if confirm != 'y':
            return

        processed_count = 0
        for file_name in text_files:
            file_path = os.path.join(folder_path, file_name)
            try:
                with open(file_path, 'r', encoding='utf-8') as f:
                    conversation = f.read().strip()

                if conversation:
                    print(f"\n🔄 Processing: {file_name}")
                    self.process_conversation(conversation, file_name)
                    processed_count += 1

                    # Add delay to avoid rate limiting
                    time.sleep(2)

            except Exception as e:
                print(f"❌ Error processing {file_name}: {e}")

        print(f"\n✅ Batch processing completed: {processed_count}/{len(text_files)} files processed")

    def process_conversation(self, conversation: str, source_name: str = "conversation"):
        """Process and analyze a conversation"""
        print("\n🔄 Processing conversation...")

        # Extract medical entities
        entities = self.assistant.extract_medical_entities(conversation)

        # Analyze conversation structure
        conversation_flow = self.assistant.analyze_conversation_structure(conversation)

        # Generate SOAP note
        patient_context = self.assistant.patient_records.get(self.current_patient, {})
        soap_note = self.assistant.generate_soap_note(entities, patient_context)

        # Create structured record
        record_id = f"{source_name}_{datetime.now().strftime('%H%M%S')}"
        structured_record = self.assistant.create_structured_record(record_id, entities, soap_note)

        # Store in patient records
        if self.current_patient not in self.assistant.patient_records:
            self.assistant.patient_records[self.current_patient] = {}
        self.assistant.patient_records[self.current_patient][record_id] = structured_record

        # Display results
        self.display_analysis_results(conversation, entities, conversation_flow, soap_note, source_name)

        # Save to file
        self.save_analysis(conversation, entities, conversation_flow, soap_note, record_id, source_name)

    def display_analysis_results(self, conversation: str, entities: Dict, conversation_flow: Dict, soap_note: str, source_name: str):
        """Display analysis results in formatted output"""
        print("\n" + "="*80)
        print("🎯 MEDICAL CONVERSATION ANALYSIS RESULTS")
        print("="*80)

        print(f"\n👤 Patient: {self.current_patient}")
        print(f"📄 Source: {source_name}")
        print(f"🕒 Analysis Time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

        # Clinical Intent
        print(f"\n🎯 CLINICAL INTENT: {entities.get('clinical_intent', 'Unknown').upper()}")

        # Symptoms
        symptoms = entities.get('symptoms', [])
        print(f"\n🧬 SYMPTOMS ({len(symptoms)} found):")
        for i, symptom in enumerate(symptoms, 1):
            print(f"   {i}. {symptom['name']}")
            if symptom.get('duration'):
                print(f"      Duration: {symptom['duration']}")
            if symptom.get('severity'):
                print(f"      Severity: {symptom['severity']}")
            if symptom.get('location'):
                print(f"      Location: {symptom['location']}")

        # Diagnoses
        diagnoses = entities.get('diagnoses', [])
        print(f"\n🏥 DIAGNOSES ({len(diagnoses)} found):")
        for i, diagnosis in enumerate(diagnoses, 1):
            print(f"   {i}. {diagnosis['name']} ({diagnosis.get('certainty', 'unknown')})")

        # Medications
        medications = entities.get('medications', [])
        print(f"\n💊 MEDICATIONS ({len(medications)} found):")
        for i, medication in enumerate(medications, 1):
            print(f"   {i}. {medication['name']}")
            if medication.get('dosage'):
                print(f"      Dosage: {medication['dosage']}")
            if medication.get('frequency'):
                print(f"      Frequency: {medication['frequency']}")
            if medication.get('purpose'):
                print(f"      Purpose: {medication['purpose']}")

        # Vital Signs
        vital_signs = entities.get('vital_signs', {})
        if any(vital_signs.values()):
            print(f"\n📊 VITAL SIGNS:")
            for key, value in vital_signs.items():
                if value:
                    print(f"   • {key.replace('_', ' ').title()}: {value}")

        # Conversation Statistics
        print(f"\n📈 CONVERSATION STATISTICS:")
        print(f"   • Total turns: {conversation_flow['total_turns']}")
        print(f"   • Doctor turns: {conversation_flow['doctor_turns']}")
        print(f"   • Patient turns: {conversation_flow['patient_turns']}")
        print(f"   • Doctor questions: {len(conversation_flow['doctor_questions'])}")

        # SOAP Note
        print(f"\n📋 GENERATED SOAP NOTE:")
        print("-" * 40)
        print(soap_note)

    def view_patient_records(self):
        """View patient records and history"""
        if self.current_patient not in self.assistant.patient_records:
            print(f"\nNo records found for {self.current_patient}")
            return

        records = self.assistant.patient_records[self.current_patient]
        print(f"\n📂 PATIENT RECORDS: {self.current_patient}")
        print(f"Total visits: {len(records)}")

        for record_id, record in records.items():
            print(f"\n📄 Record: {record_id}")
            print(f"   Date: {record['timestamp'][:10]}")
            print(f"   Intent: {record['clinical_intent']}")
            print(f"   Symptoms: {len(record['problems']['symptoms'])}")
            print(f"   Diagnoses: {len(record['problems']['diagnoses'])}")
            print(f"   Medications: {len(record['treatments']['medications'])}")

    def show_clinical_analytics(self):
        """Show clinical analytics and insights"""
        if not self.assistant.patient_records:
            print("\nNo patient data available for analytics")
            return

        insights = self.assistant.generate_clinical_insights(
            self.assistant.patient_records.get(self.current_patient, {})
        )

        print("\n📊 CLINICAL ANALYTICS")
        print("="*50)

        print(f"\n📈 VISIT STATISTICS:")
        stats = insights['visit_statistics']
        print(f"   • Total visits: {stats['total_visits']}")
        print(f"   • Average symptoms per visit: {stats['average_symptoms_per_visit']:.1f}")
        print(f"   • Common visit types: {dict(stats['common_intents'].most_common(3))}")

        print(f"\n🧬 TOP SYMPTOMS:")
        for symptom, count in list(insights['symptom_patterns'].items())[:5]:
            print(f"   • {symptom}: {count} occurrences")

        print(f"\n🏥 COMMON DIAGNOSES:")
        for diagnosis, count in list(insights['common_diagnoses'].items())[:5]:
            print(f"   • {diagnosis}: {count} occurrences")

        print(f"\n💊 FREQUENT MEDICATIONS:")
        for medication, count in list(insights['frequent_medications'].items())[:5]:
            print(f"   • {medication}: {count} occurrences")

        # Generate visualization
        self.plot_analytics(insights)

    def plot_analytics(self, insights: Dict):
        """Plot clinical analytics"""
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 12))

        # Symptoms chart
        symptoms = list(insights['symptom_patterns'].items())[:8]
        if symptoms:
            ax1.barh([s[0] for s in symptoms], [s[1] for s in symptoms], color='skyblue')
            ax1.set_title('Top Symptoms')
            ax1.set_xlabel('Frequency')

        # Diagnoses chart
        diagnoses = list(insights['common_diagnoses'].items())[:8]
        if diagnoses:
            ax2.barh([d[0] for d in diagnoses], [d[1] for d in diagnoses], color='lightcoral')
            ax2.set_title('Common Diagnoses')
            ax2.set_xlabel('Frequency')

        # Medications chart
        medications = list(insights['frequent_medications'].items())[:8]
        if medications:
            ax3.barh([m[0] for m in medications], [m[1] for m in medications], color='lightgreen')
            ax3.set_title('Frequent Medications')
            ax3.set_xlabel('Frequency')

        # Visit statistics
        stats = insights['visit_statistics']
        metrics = ['Total Visits', 'Avg Symptoms/Visit']
        values = [stats['total_visits'], stats['average_symptoms_per_visit']]
        ax4.bar(metrics, values, color=['orange', 'purple'])
        ax4.set_title('Visit Statistics')
        ax4.set_ylabel('Count')

        plt.tight_layout()
        plt.show()

    def save_analysis(self, conversation: str, entities: Dict, conversation_flow: Dict, soap_note: str, record_id: str, source_name: str):
        """Save analysis results to file"""
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        filename = f"medical_analysis_{self.current_patient}_{timestamp}.json"

        output_data = {
            "patient_id": self.current_patient,
            "record_id": record_id,
            "source_file": source_name,
            "timestamp": datetime.now().isoformat(),
            "conversation": conversation,
            "extracted_entities": entities,
            "conversation_flow": conversation_flow,
            "soap_note": soap_note
        }

        try:
            with open(filename, 'w', encoding='utf-8') as f:
                json.dump(output_data, f, indent=2, ensure_ascii=False)
            print(f"\n💾 Analysis saved to: {filename}")

            # Also save SOAP note as separate text file
            soap_filename = f"soap_note_{self.current_patient}_{timestamp}.txt"
            with open(soap_filename, 'w', encoding='utf-8') as f:
                f.write(soap_note)
            print(f"💾 SOAP note saved to: {soap_filename}")

        except Exception as e:
            print(f"Error saving analysis: {e}")

def main():
    """Main function to run the medical assistant"""
    print("🏥 Medical SLU Assistant - Clinical Dialogue Analysis System")
    print("="*60)

    # Get API key
    api_key = input("Enter your Gemini API key: ").strip()
    if not api_key:
        print("API key is required. Get one from: https://aistudio.google.com/")
        return

    try:
        # Initialize medical assistant
        medical_assistant = MedicalSLUAssistant(api_key)

        # Start user interface
        interface = MedicalAssistantInterface(medical_assistant)
        interface.start_session()

    except Exception as e:
        print(f"Error initializing medical assistant: {e}")

if __name__ == "__main__":
    main()
````

````python
import re
import json
from pathlib import Path
from typing import List, Dict, Tuple
from dataclasses import dataclass
from collections import defaultdict

@dataclass
class ConversationTurn:
    """Represents a single turn in the conversation"""
    speaker: str  # 'doctor' or 'patient'
    text: str
    turn_id: int

@dataclass
class ClinicalEntity:
    """Extracted clinical entity"""
    text: str
    entity_type: str  # 'SYMPTOM', 'DIAGNOSIS', 'MEDICATION', 'TEST', 'BODY_PART'
    start_idx: int
    end_idx: int
    turn_id: int
    confidence: float = 1.0

class ConversationParser:
    """Parse patient-doctor conversation text files"""

    def __init__(self):
        # Common patterns for speaker identification
        self.speaker_patterns = [
            (r'^(Doctor|Dr\.?|Physician):\s*(.+)$', 'doctor'),
            (r'^(Patient|Pt\.?):\s*(.+)$', 'patient'),
            (r'^D:\s*(.+)$', 'doctor'),
            (r'^P:\s*(.+)$', 'patient'),
        ]

    def parse_file(self, filepath: str) -> List[ConversationTurn]:
        """Parse a conversation file into turns"""
        with open(filepath, 'r', encoding='utf-8') as f:
            content = f.read()

        turns = []
        lines = content.strip().split('\n')
        turn_id = 0

        for line in lines:
            line = line.strip()
            if not line:
                continue

            # Try to match speaker patterns
            speaker, text = self._identify_speaker(line)

            if text:  # Valid turn found
                turns.append(ConversationTurn(
                    speaker=speaker,
                    text=text,
                    turn_id=turn_id
                ))
                turn_id += 1

        return turns

    def _identify_speaker(self, line: str) -> Tuple[str, str]:
        """Identify speaker and extract text from a line"""
        for pattern, speaker_type in self.speaker_patterns:
            match = re.match(pattern, line, re.IGNORECASE)
            if match:
                # Extract the text (last group in the match)
                text = match.group(match.lastindex).strip()
                return speaker_type, text

        # If no pattern matches, return as unknown
        return 'unknown', line

class ClinicalNER:
    """Extract clinical entities from conversations"""

    def __init__(self):
        # Medical keywords dictionaries (simplified - expand with medical ontologies)
        self.symptom_keywords = [
            'pain', 'ache', 'fever', 'cough', 'headache', 'nausea',
            'vomiting', 'dizzy', 'tired', 'fatigue', 'weakness',
            'shortness of breath', 'chest pain', 'sore throat',
            'runny nose', 'congestion', 'swelling', 'rash', 'itch'
        ]

        self.medication_keywords = [
            'aspirin', 'ibuprofen', 'acetaminophen', 'tylenol', 'advil',
            'antibiotic', 'penicillin', 'amoxicillin', 'insulin',
            'metformin', 'lisinopril', 'atorvastatin', 'prescribe',
            'medication', 'pills', 'tablet', 'mg', 'dose'
        ]

        self.diagnosis_keywords = [
            'diabetes', 'hypertension', 'infection', 'flu', 'cold',
            'pneumonia', 'bronchitis', 'diagnosis', 'condition',
            'disease', 'disorder', 'syndrome'
        ]

        self.test_keywords = [
            'blood test', 'x-ray', 'mri', 'ct scan', 'ultrasound',
            'lab work', 'test', 'screening', 'examination', 'check'
        ]

    def extract_entities(self, turns: List[ConversationTurn]) -> List[ClinicalEntity]:
        """Extract clinical entities from conversation turns"""
        entities = []

        for turn in turns:
            text_lower = turn.text.lower()

            # Extract symptoms
            entities.extend(self._extract_by_keywords(
                turn, text_lower, self.symptom_keywords, 'SYMPTOM'
            ))

            # Extract medications
            entities.extend(self._extract_by_keywords(
                turn, text_lower, self.medication_keywords, 'MEDICATION'
            ))

            # Extract diagnoses
            entities.extend(self._extract_by_keywords(
                turn, text_lower, self.diagnosis_keywords, 'DIAGNOSIS'
            ))

            # Extract tests
            entities.extend(self._extract_by_keywords(
                turn, text_lower, self.test_keywords, 'TEST'
            ))

        return entities

    def _extract_by_keywords(self, turn: ConversationTurn, text_lower: str,
                            keywords: List[str], entity_type: str) -> List[ClinicalEntity]:
        """Extract entities based on keyword matching"""
        entities = []

        for keyword in keywords:
            if keyword in text_lower:
                # Find all occurrences
                start = 0
                while True:
                    idx = text_lower.find(keyword, start)
                    if idx == -1:
                        break

                    # Extract surrounding context (for better entity boundaries)
                    context_start = max(0, idx - 20)
                    context_end = min(len(turn.text), idx + len(keyword) + 20)

                    entities.append(ClinicalEntity(
                        text=keyword,
                        entity_type=entity_type,
                        start_idx=idx,
                        end_idx=idx + len(keyword),
                        turn_id=turn.turn_id,
                        confidence=0.7  # Rule-based confidence
                    ))

                    start = idx + 1

        return entities

class ConversationSummarizer:
    """Generate structured summaries from conversations"""

    def summarize(self, turns: List[ConversationTurn],
                  entities: List[ClinicalEntity]) -> Dict:
        """Create structured summary of the conversation"""

        # Group entities by type
        entities_by_type = defaultdict(list)
        for entity in entities:
            entities_by_type[entity.entity_type].append(entity.text)

        # Remove duplicates
        for entity_type in entities_by_type:
            entities_by_type[entity_type] = list(set(entities_by_type[entity_type]))

        # Extract chief complaint (usually in first patient turn)
        chief_complaint = ""
        for turn in turns:
            if turn.speaker == 'patient':
                chief_complaint = turn.text
                break

        summary = {
            'chief_complaint': chief_complaint,
            'symptoms': entities_by_type.get('SYMPTOM', []),
            'diagnoses': entities_by_type.get('DIAGNOSIS', []),
            'medications': entities_by_type.get('MEDICATION', []),
            'tests_ordered': entities_by_type.get('TEST', []),
            'num_turns': len(turns),
            'conversation_length': sum(len(t.text) for t in turns)
        }

        return summary

# Main Pipeline
class PatientDoctorSLU:
    """Complete SLU pipeline for patient-doctor conversations"""

    def __init__(self):
        self.parser = ConversationParser()
        self.ner = ClinicalNER()
        self.summarizer = ConversationSummarizer()

    def process_file(self, filepath: str) -> Dict:
        """Process a single conversation file"""
        print(f"Processing: {filepath}")

        # Parse conversation
        turns = self.parser.parse_file(filepath)
        print(f"  Found {len(turns)} conversation turns")

        # Extract entities
        entities = self.ner.extract_entities(turns)
        print(f"  Extracted {len(entities)} clinical entities")

        # Generate summary
        summary = self.summarizer.summarize(turns, entities)

        return {
            'file': filepath,
            'turns': [vars(t) for t in turns],
            'entities': [vars(e) for e in entities],
            'summary': summary
        }

    def process_directory(self, directory: str, output_file: str = None):
        """Process all text files in a directory"""
        path = Path(directory)
        results = []

        text_files = list(path.glob('*.txt'))
        print(f"Found {len(text_files)} text files to process\n")

        for filepath in text_files:
            try:
                result = self.process_file(str(filepath))
                results.append(result)
            except Exception as e:
                print(f"Error processing {filepath}: {e}")

        # Save results
        if output_file:
            with open(output_file, 'w', encoding='utf-8') as f:
                json.dump(results, f, indent=2)
            print(f"\nResults saved to {output_file}")

        return results

# Example Usage
if __name__ == "__main__":
    # Initialize pipeline
    slu = PatientDoctorSLU()

    # Example: Process a single file
    result = slu.process_file('/content/clinical_data/RES0211.txt')
    print(json.dumps(result['summary'], indent=2))

    # Example: Process entire directory
    # results = slu.process_directory('conversations/', 'output_results.json')

    # Print sample output format
    #print("Sample Expected Output Format:")
    #print(json.dumps({
    #    'file': 'conversation_001.txt',
    #    'summary': {
    #        'chief_complaint': 'Patient reports persistent headache...',
    #        'symptoms': ['headache', 'nausea', 'dizziness'],
    #        'diagnoses': ['migraine'],
    #        'medications': ['ibuprofen', 'prescription'],
    #        'tests_ordered': ['blood test'],
    #        'num_turns': 12,
    #        'conversation_length': 1543
    #    }
    #}, indent=2))
````

````python
import os
import re
import json
import time
import warnings
from datetime import datetime
import google.generativeai as genai

warnings.filterwarnings("ignore")

# =====================================================
# INITIAL SETUP
# =====================================================
print("🏥 Medical SLU Assistant - Gemini Clinical Dialogue Analyzer")
print("============================================================")

api_key = input("🔑 Enter your Gemini API key: ").strip()
if not api_key:
    raise ValueError("API key is required. Get one from https://aistudio.google.com/")

# Configure Gemini model
genai.configure(api_key=api_key)
model = genai.GenerativeModel("gemini-2.5-flash")

# Input & output directories
DATA_DIR = "/content/clinical_data"      # Folder containing conversation .txt files
SOAP_DIR = "/content/SOAP_Notes"         # Folder to store SOAP .txt files
JSON_DIR = "/content/JSON_Results"       # Folder to store JSON files

# Create folders if not exist
os.makedirs(DATA_DIR, exist_ok=True)
os.makedirs(SOAP_DIR, exist_ok=True)
os.makedirs(JSON_DIR, exist_ok=True)

# =====================================================
# SAFE FILE READER
# =====================================================
def safe_read_text(file_path):
    """Safely read text with automatic encoding fallback."""
    encodings = ["utf-8", "latin-1", "ISO-8859-1"]
    for enc in encodings:
        try:
            with open(file_path, "r", encoding=enc) as f:
                return f.read().strip()
        except UnicodeDecodeError:
            continue
    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        return f.read().strip()

# =====================================================
# ENTITY EXTRACTION (Gemini)
# =====================================================
def extract_medical_entities(text: str) -> dict:
    """Use Gemini to extract symptoms, diagnoses, medications, etc."""
    prompt = f"""
    You are a clinical SLU specialist. Extract structured medical information from this conversation.

    DIALOGUE:
    {text}

    Return ONLY valid JSON with this structure:
    {{
      "clinical_intent": "intent_category",
      "symptoms": [{{"name": "...", "duration": "...", "severity": "...", "location": "...", "timing": "..."}}],
      "diagnoses": [{{"name": "...", "certainty": "...", "status": "..."}}],
      "medications": [{{"name": "...", "dosage": "...", "frequency": "...", "route": "...", "purpose": "...", "status": "..."}}],
      "vital_signs": {{"blood_pressure": "...", "heart_rate": "...", "temperature": "...", "weight": "..."}},
      "recommendations": [{{"type": "...", "description": "...", "timeline": "..."}}],
      "conversation_summary": "short summary"
    }}
    """

    try:
        response = model.generate_content(prompt)
        text_output = response.text.strip()
        if "```json" in text_output:
            text_output = text_output.split("```json")[1].split("```")[0]
        elif "```" in text_output:
            text_output = text_output.split("```")[1]
        return json.loads(text_output)
    except Exception as e:
        print(f"⚠️ Gemini extraction failed: {e}")
        return {
            "clinical_intent": "unknown",
            "symptoms": [],
            "diagnoses": [],
            "medications": [],
            "vital_signs": {},
            "recommendations": [],
            "conversation_summary": "fallback used"
        }

# =====================================================
# SOAP NOTE GENERATION
# =====================================================
def generate_soap(entities: dict) -> str:
    """Generate SOAP note in exact medical format."""
    soap_prompt = f"""
    Based on the extracted medical data, generate a clinical SOAP note with EXACT formatting:

    **SOAP Note:**

    ---

    **Subjective:**
    [Paragraph summarizing symptoms, duration, severity, etc.]

    ---

    **Objective:**
    [Bullet points (with hyphens) listing findings, vitals, and exam notes.]

    ---

    **Assessment:**
    [Numbered list of diagnoses.]

    ---

    **Plan:**
    [Numbered list of treatments, tests, and follow-ups.]

    DATA:
    {json.dumps(entities, indent=2)}
    """

    try:
        response = model.generate_content(soap_prompt)
        return response.text.strip()
    except Exception as e:
        return f"Error generating SOAP note: {e}"

# =====================================================
# FORMATTED OUTPUT (SYMPTOMS + DIAGNOSES)
# =====================================================
def format_symptom_list(entities: dict) -> str:
    """Return formatted list of symptoms."""
    symptoms = entities.get("symptoms", [])
    if not symptoms:
        return "No symptoms identified."

    result = "**List of symptoms:**\n"
    for s in symptoms:
        result += f"- {s.get('name', '')}"
        if s.get("location"):
            result += f" ({s['location']})"
        if s.get("severity"):
            result += f" - {s['severity']}"
        result += "\n"
    return result

def format_diagnosis_list(entities: dict) -> str:
    """Return formatted list of possible diagnoses."""
    diagnoses = entities.get("diagnoses", [])
    if not diagnoses:
        return "No diagnoses identified."

    result = "\n**List of possible diseases or illnesses:**\n"
    for d in diagnoses:
        result += f"- {d.get('name', '')}"
        if d.get("certainty") and d["certainty"] != "confirmed":
            result += f" ({d['certainty']})"
        result += "\n"
    return result

# =====================================================
# PROCESS EACH CONVERSATION FILE
# =====================================================
def process_conversation(file_path: str):
    """Process one conversation file: extract entities and generate SOAP note."""
    file_name = os.path.basename(file_path)
    base_name = os.path.splitext(file_name)[0]  # e.g., "case1"
    print(f"\n🩺 Processing: {file_name}")

    # Read conversation safely
    conversation = safe_read_text(file_path)
    if not conversation:
        print("⚠️ File is empty. Skipping.")
        return

    # 1️⃣ Extract entities
    entities = extract_medical_entities(conversation)

    # 2️⃣ Generate SOAP note
    soap_note = generate_soap(entities)

    # 3️⃣ Format sections
    symptoms_text = format_symptom_list(entities)
    diagnoses_text = format_diagnosis_list(entities)
    combined_output = f"{symptoms_text}\n{diagnoses_text}\n\n{soap_note}"

    # 4️⃣ Save files (same base name)
    soap_file_path = os.path.join(SOAP_DIR, f"{base_name}.txt")
    json_file_path = os.path.join(JSON_DIR, f"{base_name}.json")

    with open(json_file_path, "w", encoding="utf-8") as jf:
        json.dump(entities, jf, indent=2, ensure_ascii=False)

    with open(soap_file_path, "w", encoding="utf-8") as sf:
        sf.write(combined_output)

    print(f"✅ JSON saved → {json_file_path}")
    print(f"✅ SOAP Note saved → {soap_file_path}\n")

# =====================================================
# BATCH PROCESSING
# =====================================================
def batch_process(folder_path: str):
    """Process all .txt conversation files."""
    files = [f for f in os.listdir(folder_path) if f.endswith(".txt")]
    if not files:
        print("⚠️ No text files found.")
        return

    print(f"\n📂 Found {len(files)} conversation file(s). Starting processing...\n")
    for f_name in files:
        process_conversation(os.path.join(folder_path, f_name))
        time.sleep(2)  # avoid API rate limits

    print("\n🏁 All conversations processed successfully!")

# =====================================================
# RUN SCRIPT
# =====================================================
if __name__ == "__main__":
    print(f"\n📁 Reading from: {DATA_DIR}")
    print(f"🗂 SOAP Notes → {SOAP_DIR}")
    print(f"🗂 JSON Results → {JSON_DIR}")
    batch_process(DATA_DIR)
````
